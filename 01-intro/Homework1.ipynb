```python
#median of the horsepower column
print(df['horsepower'].median())
```

```python
# most frequent value in the horse power column
most_freq_val = df['horsepower'].mode()[0]
print(most_freq_val)
```

```python
# fill the horse power column with the most frequent value in the column
df['horsepower'] = df['horsepower'].fillna(most_freq_val)
```

```python
# check the column again
df['horsepower'].isnull().sum()
```

```python
# calculate the median value of horsepower again

print(df['horsepower'].median())
```

```python
# select all cars from Asia
asia_cars = df.loc[df['origin'] == 'Asia', ['vehicle_weight', 'model_year']].head(7)
```

```python
asia_cars
```

```python
X = asia_cars.values
X
```

```python
X.shape
```

```python
XTX = X.T @ X
```

```python
XTX
```

```python
XTX.shape
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
XTX_inv
```

```python
[1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
# create an array y

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```
